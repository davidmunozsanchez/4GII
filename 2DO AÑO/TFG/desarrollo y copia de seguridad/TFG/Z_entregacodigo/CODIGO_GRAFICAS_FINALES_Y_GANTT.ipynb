# GRÁFICAS FINALES

A continuación, se exponen las gráficas finales. Es necesario poner en el orden que se determinen las etiquetas del eje x, los valores de accuracy y f1. Para la std, se alternan, es decir, valor std_acc_1, valor std_f1_1, valor std_acc_2, valor std_f1_2 ...

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Datos
examples = ['NO TL NO OC METRICS', 'NO TL NO OC STD', 'SÍ TL NO OC METRICS', 'SÍ TL NO OC STD']  # Números del 1 al 6
accuracy = [0.797, 0.825, 0.79,  0.782]
accuracy_std = [0.045, 0.031, 0.012, 0.020, 0.052, 0.034, 0.018, 0.024]
f1 = [0.7824, 0.815, 0.764, 0.762]

print((np.array(accuracy) + np.array(f1))/2)
# Crear un DataFrame
df = pd.DataFrame({
    'Example': examples * 2,
    'Metric': ['Accuracy'] * len(examples) + ['F1'] * len(examples),
    'Value': accuracy + f1,
    'Std': accuracy_std,

})


# Configurar el tamaño del gráfico
plt.figure(figsize=(10, 6))

# Crear el gráfico de barras
ax = sns.barplot(x='Example', y='Value', hue='Metric', data=df, capsize=0.1)

# Añadir barras de error
for i, example in enumerate(examples):
    # Índices para las barras de 'Accuracy' y 'F1' por separado
    acc_index = i
    f1_index = i + len(examples)

    # Extraer los valores y errores estándar para 'Accuracy' y 'F1'
    acc_value = df.iloc[acc_index]['Value']
    f1_value = df.iloc[f1_index]['Value']
    acc_std = df.iloc[acc_index]['Std']
    f1_std = df.iloc[f1_index]['Std']

    # Añadir barra de error para 'Accuracy'
    plt.errorbar(x=i - 0.2, y=acc_value, yerr=acc_std, fmt='none', c='black', capsize=5, capthick=1, elinewidth=1)

    # Añadir barra de error para 'F1'
    plt.errorbar(x=i + 0.2, y=f1_value, yerr=f1_std, fmt='none', c='black', capsize=5, capthick=1, elinewidth=1)


# Configurar el título y las etiquetas
plt.title('Accuracy y F1 Score 4 IMÁGENES')
plt.xlabel('Forma de entrenar')
plt.ylabel('%')
plt.ylim(0, 1)
plt.legend(title='Métrica')

# Mostrar el gráfico
plt.show()
```

# DIAGRAMAS DE GANTT

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Datos de las tareas
df = pd.DataFrame([
    {"Task": "Preprocesado", "Start": '2024-01-20', "Finish": '2024-01-21'},
    {"Task": f"Creación de los conjuntos de datos$^1$", "Start": '2024-01-22', "Finish": '2024-02-10'},
    {"Task": "Implementación bucles entrenamiento", "Start": '2024-02-11', "Finish": '2024-02-25'},
    {"Task": f"Grid Search (30 ejecuciones)$^2$", "Start": '2024-02-26', "Finish": '2024-03-25'},
    {"Task": f"Optimización (100/75 ejecuciones)$^3$", "Start": '2024-03-26', "Finish": '2024-04-26'},
    {"Task": "Confección de la memoria", "Start": '2024-04-27', "Finish": '2024-05-27'}
])

# Convertir columnas Start y Finish a tipo datetime
df["Start"] = pd.to_datetime(df["Start"])
df["Finish"] = pd.to_datetime(df["Finish"])

# Obtener las fechas únicas de inicio y fin de cada tarea
dates = pd.concat([df["Start"], df["Finish"]]).unique()
# dates.sort()

# Colores para las tareas
colors = plt.cm.tab10.colors[:len(df)]

# Crear la figura y los ejes
fig, ax = plt.subplots(figsize=(10, 5))

# Iterar sobre las filas del dataframe para dibujar las barras de Gantt y marcar el inicio y fin de cada tarea
for i, (task, start, finish) in enumerate(zip(df["Task"], df["Start"], df["Finish"])):
    ax.barh(task, finish - start, left=start, color=colors[i], edgecolor='black')

# Configurar el eje x para mostrar solo las fechas de inicio y fin de cada tarea
ax.set_xticks(dates)
ax.set_xticklabels([date.strftime('%Y-%m-%d') for date in dates], rotation=45, ha='right')

# Etiquetas y título
ax.set_xlabel("Fecha")
ax.set_ylabel("Tarea")
ax.set_title("Diagrama de Gantt")

plt.tight_layout()  # Ajustar el diseño
plt.grid(True)  # Activar las líneas de la cuadrícula
plt.show()


```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Datos de las tareas
df = pd.DataFrame([
    {"Task": "Preprocesado", "Start": '2024-01-20', "Finish": '2024-01-21'},
    {"Task": f"Creación de los conjuntos de datos$^1$", "Start": '2024-01-22', "Finish": '2024-02-10'},
    {"Task": "Implementación bucles entrenamiento", "Start": '2024-02-11', "Finish": '2024-02-25'},
    {"Task": f"Grid Search (30 ejecuciones)$^2$", "Start": '2024-02-26', "Finish": '2024-03-25'},
    {"Task": f"Optimización (100/75 ejecuciones)$^3$", "Start": '2024-03-26', "Finish": '2024-04-26'},
    {"Task": "Confección de la memoria", "Start": '2024-04-27', "Finish": '2024-05-27'}
])

# Convertir columnas Start y Finish a tipo datetime
df["Start"] = pd.to_datetime(df["Start"])
df["Finish"] = pd.to_datetime(df["Finish"])

# Obtener la fecha de inicio de cada tarea y la fecha de finalización de la última tarea
start_dates = df["Start"].tolist()
end_date = df["Finish"].max()

# Colores para las tareas
colors = plt.cm.tab10.colors[:len(df)]

# Crear la figura y los ejes
fig, ax = plt.subplots(figsize=(65,25))

# Iterar sobre las filas del dataframe para dibujar las barras de Gantt y marcar el inicio de cada tarea
for i, (task, start, finish) in enumerate(zip(df["Task"], df["Start"], df["Finish"])):
    ax.barh(task, finish - start, left=start, color=colors[i], edgecolor='black')
    if i == len(df) - 1:  # Si es la última tarea, marcar la fecha de finalización
        ax.axvline(x=finish, color='black', linestyle='--', linewidth=0.5)

# Configurar el eje x para mostrar solo la fecha de inicio de cada tarea y la fecha de finalización de la última tarea
ax.set_xticks(start_dates + [end_date])
ax.set_xticklabels([date.strftime('%d-%m-%Y') for date in start_dates] + [end_date.strftime('%d-%m-%Y')], rotation=45, ha='right', fontsize=60)
ax.set_yticklabels(df["Task"], fontsize=60)
# Etiquetas y título
ax.set_xlabel("Fecha", fontsize=50)
ax.set_ylabel("Tarea", fontsize=50)
ax.set_title("Diagrama de Gantt", fontsize=80)

plt.tight_layout()  # Ajustar el diseño
plt.grid(True)  # Activar las líneas de la cuadrícula
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define la función max(0, x)
def max_zero_x(x):
    return np.maximum(0, x)

# Crea un rango de valores para x
x = np.linspace(-10, 10, 400)

# Calcula los valores de y
y = max_zero_x(x)

# Crea la gráfica
plt.plot(x, y, label=r'$\max(0, x)$')
plt.xlabel('x')
plt.ylabel('y')
plt.title(r'ReLU')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.legend()
plt.show()
```
